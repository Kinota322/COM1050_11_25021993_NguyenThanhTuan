```python
#1
def merge(list1, list2):
    temp_list = []
    n = 0; m = 0
    while n < len(list1) or m < len(list2):
        if n != len(list1) and (m == len(list2) or list1[n] < list2[m]):
            temp_list.append(list1[n])
            n += 1
        else:
            temp_list.append(list2[m])
            m += 1
    return temp_list

a = list(map(int, input().split()))
b = list(map(int, input().split()))

print(merge(a, b))
```

```python
#3
num = input()

prv = [0]

def upper_bound(l, r, t):
    if l == r:
        if t < int(num[prv[l]]):
            # print(t, int(num[prv[l]]))
            return prv[l]
        else:
            return None
    else:
        m = (l + r) // 2
        if int(num[prv[m]]) <= t:
            return upper_bound(m + 1, r, t)
        else:
            return upper_bound(l, m, t)

def swapped(i, j):
    new_s = ''
    for id, c in enumerate(num):
        if id == i:
            new_s += num[j]
        elif id == j:
            new_s += num[i]
        else:
            new_s += c
    return new_s

res = swapped(1, 0)

for i, j in enumerate(num):
    if i == 0: 
        continue
    temp = upper_bound(0, len(prv) - 1, int(j))
    if temp == None:
        prv.append(i)
        res = min(int(res), int(swapped(i, i - 1)))
    else:
        res = min(int(res), int(swapped(i, temp)))
        
    # print(j, upper_bound(0, len(prv) - 1, int(j)))

print(res)
```

```python
#9
def Median(nums):
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]

    if a > b: a, b = b, a

    if c > d: c, d = d, c

    if a > c:
        a, c = c, a
        b, d = d, b

    if b > e:
      b, e = e, b

    if b > c:
        return b if b < d else d
    else:
        return c if c < e else e
a=list(map(int,input().split()))
print(Median(a))
```

```python
#10
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()

    max_radius = 0
    n = len(heaters)

    for house in houses:
        idx = bisect.bisect_left(heaters, house)
        dist_left = float('inf')
        dist_right = float('inf')
        if idx < n:
            dist_right = heaters[idx] - house
        if idx > 0:
            dist_left = house - heaters[idx - 1]
        shortest_to_this_house = min(dist_left, dist_right)
        max_radius = max(max_radius, shortest_to_this_house)

    return max_radius
if __name__ == "__main__":
    house = list(map(int,input().split()))
    heat = list(map(int,input().split()))
    print(findRadius(house,heat))
```
