```python
#3
try:
    a = int(input())
except ValueError as noti:
    print(ValueError)

D = {}

def next(num):
    # print(num)
    global D
    if num == 1:
        return 'Yes'
    
    if D.get(num) == None:
        D[num] = 1
    else:
        return 'No'
    
    temp = 0
    while num:
        temp += (num % 10) ** 2
        num //= 10

    return next(temp)

print(next(a))
    
```
