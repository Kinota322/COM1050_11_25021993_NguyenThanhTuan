```python
#1
def merge(list1, list2):
    temp_list = []
    n = 0; m = 0
    while n < len(list1) or m < len(list2):
        if n != len(list1) and (m == len(list2) or list1[n] < list2[m]):
            temp_list.append(list1[n])
            n += 1
        else:
            temp_list.append(list2[m])
            m += 1
    return temp_list

a = list(map(int, input().split()))
b = list(map(int, input().split()))

print(merge(a, b))
```

```python
#3
num = input()

prv = [0]

def upper_bound(l, r, t):
    if l == r:
        if t < int(num[prv[l]]):
            # print(t, int(num[prv[l]]))
            return prv[l]
        else:
            return None
    else:
        m = (l + r) // 2
        if int(num[prv[m]]) <= t:
            return upper_bound(m + 1, r, t)
        else:
            return upper_bound(l, m, t)

def swapped(i, j):
    new_s = ''
    for id, c in enumerate(num):
        if id == i:
            new_s += num[j]
        elif id == j:
            new_s += num[i]
        else:
            new_s += c
    return new_s

res = swapped(1, 0)

for i, j in enumerate(num):
    if i == 0: 
        continue
    temp = upper_bound(0, len(prv) - 1, int(j))
    if temp == None:
        prv.append(i)
        res = min(int(res), int(swapped(i, i - 1)))
    else:
        res = min(int(res), int(swapped(i, temp)))
        
    # print(j, upper_bound(0, len(prv) - 1, int(j)))

print(res)
```

```python
#5
a = list(map(int, input().split()))
b = list(map(int, input().split()))

res = False

def check(j):
    # print(a, b[j:(j + len(a))])
    if a == b[j:(j + len(a))]:
        return True
    else:
        return False

for i in range(0, len(b) - len(a)):
    if a[0] == b[i] and check(i):
        res = True
        break

print(res)
```
