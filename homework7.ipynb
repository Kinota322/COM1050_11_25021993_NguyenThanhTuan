```python
#1
def binary_search(arr, left, right, target):
    # print(arr, left, right, target)
    if left == right:
        if arr[left] == target:
            return left
        else:
            return -1
    else:
        mid = (left + right) // 2
        if arr[mid] < target:
            return binary_search(arr, mid + 1, right, target)
        else:
            return binary_search(arr, left, mid, target)

a = list(map(int, input().split()))
k = int(input())

print(binary_search(a, 0, len(a) - 1, k))
```

```python
#2
def count_occurrences(arr, x):
    cnt = 0
    for i in arr:  
        cnt += (i == x)
    return cnt

a = list(map(int, input().split()))
k = int(input())

print(count_occurrences(a, k))
```

```python
#3
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(0, len(arr) - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

def selection_sort(arr):
    for i in range(len(arr)):
        mn = i
        for j in range(i + 1, len(arr)):
            if arr[j] < arr[mn]:
                mn = j
        arr[i], arr[mn] = arr[mn], arr[i]

a = list(map(int, input().split()))

# bubble_sort(a)

selection_sort(a)

print(a)
```

```python
#9
arr = list(map(int, input().split()))

node = [[0, 0]] * 1000000

def build_tree(i, l, r):
    # global node

    # print(i, l, r)
    if l == r:
        # print('---', i, l, r)
        node[i] = [arr[l], arr[l]]#min
        # print(node)
    else:
        m = (l + r) // 2
        build_tree(i * 2, l, m)
        build_tree(i * 2 + 1, m + 1, r)

        # print(i, l, r, i * 2, i * 2 + 1, node[i * 2], node[i * 2 + 1])

        node[i] = [min(node[i * 2][0], node[i * 2 + 1][0]), max(node[i * 2][1], node[i * 2 + 1][1])]
        # return 0

    # print(i, l, r, node[i])

def smaller(i, l, r, t, t_l, t_r):
    if t_l <= l and t_r >= r:
        if node[i][0] < t and node[i][1] < t:
            return r - l + 1
        elif node[i][0] < t and node[i][1] >= t:
            m = (l + r) // 2
            return smaller(i * 2, l, m, t, t_l, t_r) + smaller(i * 2 + 1, m + 1, r, t, t_l, t_r)
        else:
            return 0
    elif l == r:
        return 0
    else:
        m = (l + r) // 2
        return smaller(i * 2, l, m, t, t_l, t_r) + smaller(i * 2 + 1, m + 1, r, t, t_l, t_r) 
    
build_tree(1, 0, len(arr) - 1)

for i, j in enumerate(arr):
    print(smaller(1, 0, len(arr) - 1, j, i + 1, len(arr) - 1), end = ' ')
```
