```python
#1

pi = 3.14

class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
    def area(self):
        return f'{((pi * (self.radius ** 2)) * 2 + (2 * pi * self.radius) * self.height):.2f}'
    def volume(self):
        return f'{((pi * (self.radius ** 2)) * self.height):.2f}'        
    
P = Cylinder(3, 3)

print(P.area(), P.volume())
```

```python
#2
class Date:
    def __init__(self, d, m, y):
        self.day = d
        self.month = m
        self.year = y
    def nam_nhuan(self):
        return self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0)

    def days_in_month(self):
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            return 31
        if self.month in [4, 6, 9, 11]:
            return 30
        if self.month == 2:
            return 29 if self.nam_nhuan() else 28
        return 0

    def is_valid(self):
        if self.year < 1 or self.month < 1 or self.month > 12:
            return False
        if self.day < 1 or self.day > self.days_in_month():
            return False
        return True

    def next_date(self):
        self.day += 1
        if self.day > self.days_in_month():
            self.day = 1
            self.month += 1
            if self.month > 12:
                self.month = 1
                self.year += 1
d, m, y = map(int, input().split())
date = Date(d, m, y)
if date.is_valid():
    date.next_date()
    print(f"{date.day}/{date.month}/{date.year}")
     
```

```python
#5
class ShoppingCart:
    def __init__(self):
        self.products = []
    def add_product(self, name, price):
        product = {"name": name, "price": price}
        self.products.append(product)
    def remove_product(self, name):
        for p in self.products:
            if p["name"] == name:
                self.products.remove(p)
                return True
        return False
    def is_empty(self):
        return len(self.products) == 0
    def total_price(self):
        total = 0
        for p in self.products:
            total += p["price"]
        return total
    def show_products(self):
        if self.is_empty():
            print("Gio hang rong")
        else:
            for p in self.products:
                print(p["name"], "-", p["price"])
    def clear_cart(self):
        self.products.clear()
cart = ShoppingCart()
while True:
    print("2. Them san pham")
    print("3. Xoa san pham theo ten")
    print("4. Kiem tra gio hang rong")
    print("5. Tinh tong tien")
    print("6. Hien thi gio hang")
    print("7. Xoa toan bo gio hang")
    choice = input("Nhap lua chon (2-7): ")
    if choice == "2":
        name = input("Nhap ten san pham: ")
        price = float(input("Nhap gia san pham: "))
        cart.add_product(name, price)
    elif choice == "3":
        name = input("Nhap ten san pham can xoa: ")
        if cart.remove_product(name):
            print("Da xoa san pham")
        else:
            print("Khong tim thay san pham")
    elif choice == "4":
        if cart.is_empty():
            print("Gio hang dang rong")
        else:
            print("Gio hang khong rong")
    elif choice == "5":
        print("Tong tien:", cart.total_price())
    elif choice == "6":
        cart.show_products()
    elif choice == "7":
        cart.clear_cart()
        print("Da xoa toan bo gio hang")
    else:
        print("Lua chon khong hop le")
    stop = input("Co muon thoat khoi chuong trinh khong? ")
    if stop == "yes":
        break
```

```python
#6
class Rectangle:
  def __init__(self, w, h):
    self.w = w
    self.h = h
  def area(self):
    return self.w * self.h
  def perimeter(self):
    return 2 * (self.w + self.h)
  def scale(self, k):
    self.w *= k
    self.h *= k
w, h, k = map(float, input().split())
rect = Rectangle(w, h)
rect.scale(k)
print(f"{rect.area():.2f} {rect.perimeter():.2f}")
```

```python
#7
from math import *
class Fraction :
    def __init__(self,num,den):
        if den < 0:
            num = - num
            den = - den
        common = gcd(num,den)
        self.num = num // common
        self.den = den // common
    def add(self,other):
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def sub(self,other):
        new_num = self.num * other.den - self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def mul(self,other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def div(self,other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num,new_den)
if __name__ == '__main__':
    a = input().split()
    t1 = int(a[0])
    m1 = int(a[1])
    op = a[2]
    t2 = int(a[3])
    m2 = int(a[4])
    s1 = Fraction(t1,m1)
    s2 = Fraction(t2,m2)
    if op == '+':
        result = s1.add(s2)
        print(f'{result.num}/{result.den}')
    elif op == '-':
        result = s1.sub(s2)
        print(f'{result.num}/{result.den}')
    elif op == '*':
        result = s1.mul(s2)
        print(f'{result.num}/{result.den}')
    elif op == '/':
        result = s1.div(s2)
        print(f'{result.num}/{result.den}')
```

```python
#8
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def deposit(self,x):
        self.balance += x
    def withdraw(self,x):
        if self.balance < x:
            self.balance = self.balance
        else :
            self.balance -= x
if __name__ == '__main__':
    a = input().split()
    bank = BankAccount(a[0],float(a[1]))
    n = int(input())
    while n > 0:
        giao_dich = input().split()
        if giao_dich[0] == 'DEPOSIT':
            bank.deposit(float(giao_dich[1]))
        elif giao_dich[0] == 'WITHDRAW':
            bank.withdraw(float(giao_dich[1]))
        n -= 1
    print(f'{bank.balance:.2f}')
```

```python
#9
class Student:
    def __init__(self,name):
        self.name = name
        self.sum = 0
        self.cnt = 0
    def add(self,subject,score):
        self.score = score
    def tong(self):
        self.sum += self.score
        self.cnt += 1
    def avg(self):
        result = self.sum / self.cnt
        return result
    def rank(self):
        if self.avg() >= 8:
            return 'Excellent'
        elif self.avg() >= 6.5:
            return 'Good'
        elif self.avg() >= 5:
            return 'Average'
        else:
            return 'Poor'
if __name__ == '__main__':
    name = input()
    n = int(input())
    hs = Student(name)
    while n > 0:
        subject_score = input().split()
        hs.add(subject_score[0],float(subject_score[1]))
        hs.tong()
        n -= 1
    print(name,f'{hs.avg():.2f}',hs.rank())
```

```python
#10
class Book:
    def __init__(self,title,author,year):
        self.title = title
        self.author = author
        self.year = year
class Library:
    def __init__(self):
        self.book = []
    def add(self,book):
        self.book.append(book)
    def count_by_author(self,name):
        cnt = 0
        for book in self.book:
            if book.author == name:
                cnt += 1
        return cnt
    def find_by_year(self,year):
        cnt = 0
        for book in self.book:
            if book.year == year:
                cnt += 1
        return cnt
if __name__ == '__main__':
    n = int(input())
    lib = Library()
    while n > 0:
        thao_tac = input().split()
        if thao_tac[0] == 'ADD':
            b = thao_tac[1].split(';')
            book = Book(b[0],b[1],b[2])
            lib.add(book)
        elif thao_tac[0] == 'COUNT':
            print(lib.count_by_author(thao_tac[1]))
        elif thao_tac[0] == 'COUNTYEAR':
            print(lib.find_by_year(thao_tac[1]))
        n -= 1
```

```python
#11
from math import *
class Angle :
    def __init__(self,a,b,c,d):
        self.ABx = b[0] - a[0]
        self.ABy = b[1] - a[1]
        self.ABz = b[2] - a[2]
        self.ACx = c[0] - a[0]
        self.ACy = c[1] - a[1]
        self.ACz = c[2] - a[2]
        self.BCx = c[0] - b[0]
        self.BCy = c[1] - b[1]
        self.BCz = c[2] - b[2]
        self.BDx = d[0] - b[0]
        self.BDy = d[1] - b[1]
        self.BDz = d[2] - b[2]
    def vector(self):
        self.ABC = [self.ABy * self.ACz - self.ABz * self.ACy, self.ABz * self.ACx - self.ABx * self.ACz, self.ABx * self.ACy - self.ABy * self.ACx]
        self.BCD = [self.BCy * self.BDz - self.BCz * self.BDy, self.BCz * self.BDx - self.BCx * self.BDz, self.BCx * self.BDy - self.BCy * self.BDx]
    def cos(self):
        tich_vo_huong = self.ABC[0] * self.BCD[0] + self.ABC[1] * self.BCD[1] + self.ABC[2] * self.BCD[2]
        tich_do_dai = ((self.ABC[0] ** 2 + self.ABC[1] ** 2 + self.ABC[2] ** 2) ** 0.5) *((self.BCD[0] ** 2 + self.BCD[1] ** 2 + self.BCD[2] ** 2) ** 0.5)
        cos = abs(tich_vo_huong/tich_do_dai)
        return cos
if __name__ == '__main__' :
    A = list(map(float,input().split()))
    B = list(map(float,input().split()))
    C = list(map(float,input().split()))
    D = list(map(float,input().split()))
    my_angle = Angle(A,B,C,D)
    my_angle.vector()
    my_angle = acos(my_angle.cos())
    my_angle = degrees(my_angle)
    print(f'{my_angle:.2f}')
```

```python
#12
class Complex:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def pro(self,other):
        self.thuc = self.a * other.a - self.b * other.b
        self.ao = self.a * other.b + self.b * other.a
if __name__ == '__main__':
    a = list(map(int,input().split()))
    b = list(map(int,input().split()))
    my_complex = Complex(a[0],a[1])
    other_complex = Complex(b[0],b[1])
    my_complex.pro(other_complex)
    print(my_complex.thuc,my_complex.ao)
```

```python
#13
class Triangle:
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c
    def getArea(self):
        if self.a + self.b <= self.c or self.b + self.c <= self.a or self.a + self.c <= self.b:
            return -1
        self.p = (self.a + self.b + self.c)/2
        return (self.p * (self.p -self.a) * (self.p - self.b) * (self.p -self.c)) ** 0.5
if __name__ == '__main__':
    my_tri = Triangle(*map(int,input().split()))
    if my_tri.getArea() == -1 :
        print('invalid')
    else:
        print(f'{my_tri.getArea():.2f}')
```
